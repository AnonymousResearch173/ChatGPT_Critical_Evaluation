```python
import pandas as pd
import re
from bs4 import BeautifulSoup

df = pd.read_csv("/content/SO_ QueryResults Java.csv")

# Function to extract & clean answer code

def extract_code(body_html):
    if pd.isna(body_html):
        return ""

    soup = BeautifulSoup(body_html, "html.parser")

    # Extract all <code> blocks
    code_blocks = soup.find_all("code")
    code_texts = []

    for block in code_blocks:
        text = block.get_text("\n", strip=True)
        code_texts.append(text)

    # Join multiple code blocks into one code section
    return "\n\n".join(code_texts)

# Function to compute LOC

def compute_loc(code_text):
    if not code_text:
        return 0

    # Split into lines, remove empty or whitespace-only lines
    lines = [line for line in code_text.split("\n") if line.strip()]
    return len(lines)

# cleaning + LOC calculation

df["Code"] = df["AnswerBody"].apply(extract_code)
df["LOC"] = df["Code"].apply(compute_loc)

# Filter rows with LOC ≥ 15

df_filtered = df[df["LOC"] >= 15].reset_index(drop=True)


df_filtered.to_csv("cleaned_filtered_answers.csv", index=False)

print("Total original rows:", len(df))
print("Rows with LOC >= 15:", len(df_filtered))
print("Saved as cleaned_filtered_answers.csv")
```
